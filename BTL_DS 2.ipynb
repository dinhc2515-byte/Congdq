```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dữ liệu ban đầu
data = {
    "ngay": [
        "2026-01-01", "2026-01-02", "2026-01-03",
        "2026-01-01", "2026-01-02", "2026-01-03",
        "2026-01-01", "2026-01-02", "2026-01-03",
        "2026-01-01", "2026-01-02", "2026-01-03"
    ],
    "quan": [
        "Quan 1", "Quan 1", "Quan 1",
        "Quan 3", "Quan 3", "Quan 3",
        "Quan 5", "Quan 5", "Quan 5",
        "Quan 7", "Quan 7", "Quan 7"
    ],
    "pm25": [35, 42, 28, 50, 60, 45, 25, 30, np.nan, 40, 38, 32],
    "pm10": [70, 80, 55, 95, 110, 90, 50, 60, 65, 85, np.nan, 70],
    "nhiet_do": [30, 31, 29, 32, 33, 31, 28, 29, 30, 31, 30, 29],
    "do_am": [65, 60, 70, 58, 55, 62, 75, 72, 68, 63, 66, 70],
    "luong_mua": [0, 0, 5, 0, 0, 3, 10, 2, 0, 0, 1, 4],
    "chat_luong": [
        "Trung binh", "Kem", "Tot",
        "Kem", "Xau", "Kem",
        "Tot", "Trung binh", "Trung binh",
        "Kem", "Trung binh", "Trung binh"
    ]
}

df = pd.DataFrame(data)

# ================= PHẦN 1 =================
pm25 = np.array([35, 42, 28, 50, 60, 45, 25, 30, 40, 38, 32])

print("Số lượng phần tử:", len(pm25))
print("Tổng pm25:", np.sum(pm25))
print("Trung bình pm25:", np.mean(pm25))
print("Lớn nhất:", np.max(pm25))
print("Nhỏ nhất:", np.min(pm25))
print("pm25 > 40:", pm25[pm25 > 40])

# ================= PHẦN 2 =================
pm10 = np.array([70, 80, 55, 95, 110, 90, 50, 60, 85, 70])

print("Trung bình:", np.mean(pm10))
print("Trung vị:", np.median(pm10))
print("Độ lệch chuẩn:", np.std(pm10))
print("Phương sai:", np.var(pm10))
print("Phân vị 25%, 50%, 75%:", np.percentile(pm10, [25, 50, 75]))
print("Vị trí lớn nhất:", np.argmax(pm10))
print("Vị trí nhỏ nhất:", np.argmin(pm10))
print("Nhận xét: Dữ liệu pm10 có sự biến động khá rõ giữa các giá trị.")

# ================= PHẦN 3 =================
data_pm25 = np.array([35, 42, np.nan, 50, 60, 45, np.nan, 30, 40])

print("Vị trí bị thiếu:", np.isnan(data_pm25))
print("Số lượng bị thiếu:", np.sum(np.isnan(data_pm25)))

tb_pm25 = np.nanmean(data_pm25)
print("Trung bình bỏ qua NaN:", tb_pm25)

data_pm25[np.isnan(data_pm25)] = tb_pm25
print("Sau khi thay NaN:", data_pm25)

phan_loai = np.where(data_pm25 < 35, "Tot", np.where(data_pm25 <= 50, "Trung binh", "Xau"))

print("Phân loại:", phan_loai)

# ================= PHẦN 4 =================
series_quan = pd.Series(["Quan 1", "Quan 3", "Quan 5", "Quan 7"])
print(series_quan)

df = pd.DataFrame(data)

print(df.head())
print(df.tail())
print(df.info())
print(df.describe())

# ================= PHẦN 5 =================
df.to_csv("air_quality.csv", index=False)

df_read = pd.read_csv("air_quality.csv")
print(df_read.head())
print("Số dòng, số cột:", df_read.shape)

df.to_excel("air_quality.xlsx", index=False)

# ================= PHẦN 6 =================
print(df[df["pm25"] > 40])
print(df[df["quan"] == "Quan 3"])
print(df[(df["pm25"] > 35) & (df["luong_mua"] == 0)])
print(df[df["chat_luong"].isin(["Kem", "Xau"])])
print(df[["ngay", "quan", "pm25", "chat_luong"]])

# ================= PHẦN 7 =================
print(df.sort_values(by="pm25", ascending=False))
print(df.sort_values(by=["quan", "ngay"]))

pm25_tb_quan = df.groupby("quan")["pm25"].mean()
print(pm25_tb_quan)

pm10_max_quan = df.groupby("quan")["pm10"].max()
print(pm10_max_quan)

print("Quận có pm25 trung bình cao nhất:", pm25_tb_quan.idxmax())

# ================= PHẦN 8 =================
print(df.isnull().sum())

df["pm25"] = df["pm25"].fillna(df["pm25"].mean())
df["pm10"] = df["pm10"].fillna(df["pm10"].median())

print(df.duplicated())

df = df.drop_duplicates()

df["ngay"] = pd.to_datetime(df["ngay"])

# ================= PHẦN 9 =================
df["chi_so_o_nhiem"] = df["pm25"] * 0.6 + df["pm10"] * 0.4

df["muc_canh_bao"] = np.where(df["chi_so_o_nhiem"] < 50, "Thap",
                     np.where(df["chi_so_o_nhiem"] < 80, "Trung binh", "Cao"))

pivot = pd.pivot_table(
    df,
    values="pm25",
    index="quan",
    columns="chat_luong",
    aggfunc="mean"
)

print(pivot)

df = df.rename(columns={
    "pm25": "PM2.5",
    "pm10": "PM10"
})

df.to_csv("air_quality_processed.csv", index=False)

# ================= PHẦN 10 =================

# Biểu đồ đường pm25 của Quan 1
quan1 = df[df["quan"] == "Quan 1"]

plt.plot(quan1["ngay"], quan1["PM2.5"], marker="o", label="Quan 1")
plt.title("Sự thay đổi PM2.5 theo ngày của Quận 1")
plt.xlabel("Ngày")
plt.ylabel("PM2.5")
plt.legend()
plt.grid()
plt.show()

# Biểu đồ cột pm25 trung bình theo quận
pm25_tb = df.groupby("quan")["PM2.5"].mean()

pm25_tb.plot(kind="bar")
plt.title("PM2.5 trung bình giữa các quận")
plt.xlabel("Quận")
plt.ylabel("PM2.5 trung bình")
plt.show()

# Biểu đồ tròn tỷ lệ chất lượng
df["chat_luong"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Tỷ lệ các mức chất lượng không khí")
plt.ylabel("")
plt.show()

# Biểu đồ scatter pm25 và pm10
plt.scatter(df["PM2.5"], df["PM10"])
plt.title("Mối quan hệ giữa PM2.5 và PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")
plt.grid()
plt.show()
```
